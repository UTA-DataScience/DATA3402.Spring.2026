1. Write a "counter" class that can be incremented up to a specified maximum value, will print an error if an attempt is made to increment beyond that value, and allows reseting the counter. 

```python
class SimpleCounter:
    def __init__(self, max_value):
        self.count = 0
        self.max_value = max_value

    def increment(self):
        if self.count >= self.max_value:
            print("Error: Attempted to increment beyond maximum value.")
        else:
            self.count += 1

    def reset(self):
        self.count = 0

# --- Test ---
c1 = SimpleCounter(2)
c1.increment()
c1.increment()
c1.increment() # This will print the error
c1.reset()
c1.increment() 
c1.increment()
c1.increment() # This will print the error again
```

```python
class Counter:
    def __init__(self, max_value):
        self.__count = 0        
        self.__max_value = max_value

    def increment(self):
        if self.__count >= self.__max_value:
            print("Error: Attempted to increment beyond maximum value.")
        else:
            self.__count += 1

    def reset(self):
        self.__count = 0

    def get_value(self):
        return self.__count

    def get_max(self):
        return self.__max_value

    def is_at_max(self):
        return self.__count == self.__max_value

c2 = Counter(3)
c2.increment()
c2.increment()
print(f"Value: {c2.get_value()}, Is Max?: {c2.is_at_max()}")
```

```python
class Shape:
    def area(self): raise NotImplementedError
    def perimeter(self): raise NotImplementedError
    def get_perimeter_points(self): raise NotImplementedError
    def contains(self, x, y): raise NotImplementedError

class Rectangle(Shape):
    def __init__(self, length, width, x, y):
        self.length = length
        self.width = width
        self.x = x
        self.y = y
        
    def area(self): return self.length * self.width
    def perimeter(self): return 2 * (self.length + self.width)
    
    def get_perimeter_points(self):
        points = []
        for i in range(4):
            points.append((self.x + (self.length * i / 4), self.y))
            points.append((self.x + (self.length * i / 4), self.y - self.width))
            points.append((self.x, self.y - (self.width * i / 4)))
            points.append((self.x + self.length, self.y - (self.width * i / 4)))
        return points

    def contains(self, px, py):
        return (self.x <= px <= self.x + self.length) and (self.y - self.width <= py <= self.y)

    def __repr__(self):
        return f"Rectangle({self.length}, {self.width}, {self.x}, {self.y})"

class Circle(Shape):
    def __init__(self, radius, cx, cy):
        self.radius = radius
        self.cx = cx
        self.cy = cy
        
    def area(self): return math.pi * (self.radius ** 2)
    def perimeter(self): return 2 * math.pi * self.radius
    
    def get_perimeter_points(self):
        points = []
        for i in range(16):
            angle = 2 * math.pi * i / 16
            points.append((self.cx + self.radius * math.cos(angle), self.cy + self.radius * math.sin(angle)))
        return points

    def contains(self, px, py):
        return math.sqrt((px - self.cx)**2 + (py - self.cy)**2) <= self.radius

    def __repr__(self):
        return f"Circle({self.radius}, {self.cx}, {self.cy})"

class Triangle(Shape):
    def __init__(self, base, height, x, y):
        self.base = base
        self.height = height
        self.x = x
        self.y = y
        
    def area(self): return 0.5 * self.base * self.height
    def perimeter(self): return self.base + self.height + math.sqrt(self.base**2 + self.height**2)
    
    def get_perimeter_points(self):
        points = []
        for i in range(5):
            points.append((self.x + (self.base * i / 5), self.y))
            points.append((self.x, self.y + (self.height * i / 5)))
            points.append((self.x + (self.base * i / 5), self.y + self.height - (self.height * i / 5)))
        points.append((self.x + self.base, self.y))
        return points

    def contains(self, px, py):
        if not ((self.x <= px <= self.x + self.base) and (self.y <= py <= self.y + self.height)):
            return False
        return py <= self.y + self.height - (self.height / self.base) * (px - self.x)

    def __repr__(self):
        return f"Triangle({self.base}, {self.height}, {self.x}, {self.y})"
```

9. Add a function in the base class of the object classes that returns true/false testing that the object overlaps with another object.

```python
class Shape:
    def area(self): raise NotImplementedError
    def perimeter(self): raise NotImplementedError
    def get_perimeter_points(self): raise NotImplementedError
    def contains(self, x, y): raise NotImplementedError
    
    def overlaps(self, other):
        for px, py in other.get_perimeter_points():
            if self.contains(px, py):
                return True
                
        for px, py in self.get_perimeter_points():
            if other.contains(px, py):
                return True
                
        return False

class Rectangle(Shape):
    def __init__(self, length, width, x, y):
        self.length = length
        self.width = width
        self.x = x
        self.y = y
        
    def area(self): return self.length * self.width
    def perimeter(self): return 2 * (self.length + self.width)
    
    def get_perimeter_points(self):
        points = []
        for i in range(4):
            points.append((self.x + (self.length * i / 4), self.y))
            points.append((self.x + (self.length * i / 4), self.y - self.width))
            points.append((self.x, self.y - (self.width * i / 4)))
            points.append((self.x + self.length, self.y - (self.width * i / 4)))
        return points

    def contains(self, px, py):
        return (self.x <= px <= self.x + self.length) and (self.y - self.width <= py <= self.y)

    def __repr__(self):
        return f"Rectangle({self.length}, {self.width}, {self.x}, {self.y})"

class Circle(Shape):
    def __init__(self, radius, cx, cy):
        self.radius = radius
        self.cx = cx
        self.cy = cy
        
    def area(self): return math.pi * (self.radius ** 2)
    def perimeter(self): return 2 * math.pi * self.radius
    
    def get_perimeter_points(self):
        points = []
        for i in range(16):
            angle = 2 * math.pi * i / 16
            points.append((self.cx + self.radius * math.cos(angle), self.cy + self.radius * math.sin(angle)))
        return points

    def contains(self, px, py):
        return math.sqrt((px - self.cx)**2 + (py - self.cy)**2) <= self.radius

    def __repr__(self):
        return f"Circle({self.radius}, {self.cx}, {self.cy})"

# --- Q9 Test ---
r1 = Rectangle(10, 10, 0, 10)
c1 = Circle(5, 5, 5)
print(r1.overlaps(c1))
```

10. Copy the `Canvas` class from lecture to in a python file creating a `paint` module. Copy your classes from above into the module and implement paint functions. Implement a `CompoundShape` class. Create a simple drawing demonstrating that all of your classes are working.

```python
import math

class Canvas:
    def __init__(self, width, height):
        self.width = width
        self.height = height
        self.data = [[' '] * width for i in range(height)]

    def set_pixel(self, row, col, char='*'):
        if 0 <= row < self.height and 0 <= col < self.width:
            self.data[row][col] = char

    def get_pixel(self, row, col):
        return self.data[row][col]
    
    def clear_canvas(self):
        self.data = [[' '] * self.width for i in range(self.height)]
        
    def display(self):
        print("\n".join(["".join(row) for row in self.data]))

class Shape:
    def __init__(self, name="", char="*"):
        self.name = name
        self.char = char
        
    def area(self): raise NotImplementedError
    def perimeter(self): raise NotImplementedError
    def get_perimeter_points(self): raise NotImplementedError
    def contains(self, x, y): raise NotImplementedError
    
    def overlaps(self, other):
        for px, py in other.get_perimeter_points():
            if self.contains(px, py): return True
        for px, py in self.get_perimeter_points():
            if other.contains(px, py): return True
        return False
        
    def paint(self, canvas):
        for row in range(canvas.height):
            for col in range(canvas.width):
                if self.contains(col, row):
                    canvas.set_pixel(row, col, self.char)

class Rectangle(Shape):
    def __init__(self, length, width, x, y, name="", char="*"):
        super().__init__(name, char)
        self.length = length
        self.width = width
        self.x = x
        self.y = y
        
    def area(self): return self.length * self.width
    def perimeter(self): return 2 * (self.length + self.width)
    
    def get_perimeter_points(self):
        points = []
        for i in range(4):
            points.append((self.x + (self.length * i / 4), self.y))
            points.append((self.x + (self.length * i / 4), self.y + self.width))
            points.append((self.x, self.y + (self.width * i / 4)))
            points.append((self.x + self.length, self.y + (self.width * i / 4)))
        return points

    def contains(self, px, py):
        return (self.x <= px <= self.x + self.length) and (self.y <= py <= self.y + self.width)

    def __repr__(self):
        return f"Rectangle({self.length}, {self.width}, {self.x}, {self.y}, '{self.name}', '{self.char}')"

class Circle(Shape):
    def __init__(self, radius, cx, cy, name="", char="*"):
        super().__init__(name, char)
        self.radius = radius
        self.cx = cx
        self.cy = cy
        
    def area(self): return math.pi * (self.radius ** 2)
    def perimeter(self): return 2 * math.pi * self.radius
    
    def get_perimeter_points(self):
        points = []
        for i in range(16):
            angle = 2 * math.pi * i / 16
            points.append((self.cx + self.radius * math.cos(angle), self.cy + self.radius * math.sin(angle)))
        return points

    def contains(self, px, py):
        return math.sqrt((px - self.cx)**2 + (py - self.cy)**2) <= self.radius

    def __repr__(self):
        return f"Circle({self.radius}, {self.cx}, {self.cy}, '{self.name}', '{self.char}')"

class Triangle(Shape):
    def __init__(self, base, height, x, y, name="", char="*"):
        super().__init__(name, char)
        self.base = base
        self.height = height
        self.x = x
        self.y = y
        
    def area(self): return 0.5 * self.base * self.height
    def perimeter(self): return self.base + self.height + math.sqrt(self.base**2 + self.height**2)
    
    def get_perimeter_points(self):
        points = []
        for i in range(5):
            points.append((self.x + (self.base * i / 5), self.y))
            points.append((self.x, self.y + (self.height * i / 5)))
            points.append((self.x + (self.base * i / 5), self.y + self.height - (self.height * i / 5)))
        points.append((self.x + self.base, self.y))
        return points

    def contains(self, px, py):
        if not ((self.x <= px <= self.x + self.base) and (self.y <= py <= self.y + self.height)):
            return False
        return py <= self.y + self.height - (self.height / self.base) * (px - self.x)

    def __repr__(self):
        return f"Triangle({self.base}, {self.height}, {self.x}, {self.y}, '{self.name}', '{self.char}')"

class CompoundShape(Shape):
    def __init__(self, shapes, name="", char="*"):
        super().__init__(name, char)
        self.shapes = shapes
        
    def get_perimeter_points(self):
        points = []
        for s in self.shapes: 
            points.extend(s.get_perimeter_points())
        return points
        
    def contains(self, x, y):
        return any(s.contains(x, y) for s in self.shapes)
        
    def paint(self, canvas):
        for s in self.shapes:
            s.paint(canvas)
            
    def __repr__(self):
        return f"CompoundShape({self.shapes}, '{self.name}', '{self.char}')"

my_canvas = Canvas(40, 20)
comp = CompoundShape([Rectangle(10, 5, 2, 2, char="#"), Circle(4, 25, 8, char="O")])
comp.paint(my_canvas)
my_canvas.display()
```

11. Create a `RasterDrawing` class. Demonstrate that you can create a drawing made of several shapes, paint the drawing, modify the drawing, and paint it again. 

```python
class RasterDrawing:
    def __init__(self):
        self.shapes = dict()
        self.shape_names = list()
        
    def add_shape(self, shape):
        if shape.name == "":
            shape.name = self.assign_name()
        self.shapes[shape.name] = shape
        self.shape_names.append(shape.name)

    def update(self, canvas):
        canvas.clear_canvas()
        self.paint(canvas)
        
    def paint(self, canvas):
        for shape_name in self.shape_names:
            self.shapes[shape_name].paint(canvas)
            
    def assign_name(self):
        name_base = "shape"
        name = name_base + "_0"
        i = 1
        while name in self.shapes:
            name = name_base + "_" + str(i)
            i += 1
        return name

drawing_canvas = Canvas(40, 20)
rd = RasterDrawing()
rd.add_shape(Circle(5, 10, 10, char="O"))
rd.add_shape(Triangle(8, 8, 20, 5, char="T"))
rd.paint(drawing_canvas)
drawing_canvas.display()

print("\n")

rd.shapes["shape_0"].radius = 8
rd.update(drawing_canvas)
drawing_canvas.display()
```

12. Implement the ability to load/save raster drawings and demonstate that your method works. One way to implement this ability:

   * Overload `__repr__` functions of all objects to return strings of the python code that would construct the object.
   
   * In the save method of raster drawing class, store the representations into the file.
   * Write a loader function that reads the file and uses `eval` to instantiate the object.

For example:

```python
class RasterDrawingSaveable(RasterDrawing):
    def save(self, filename):
        with open(filename, "w") as f:
            f.write(repr(list(self.shapes.values())))
            
def load_drawing(filename):
    with open(filename, "r") as f:
        shapes_list = eval(f.read())
        
    new_drawing = RasterDrawingSaveable()
    for s in shapes_list:
        new_drawing.add_shape(s)
    return new_drawing

drawing = RasterDrawingSaveable()
drawing.add_shape(Rectangle(10, 5, 5, 5, char="#"))
drawing.add_shape(Circle(3, 25, 10, char="O"))

drawing.save("lab4_drawing.txt")

loaded = load_drawing("lab4_drawing.txt")
test_canvas = Canvas(40, 20)
loaded.paint(test_canvas)
test_canvas.display()
```
