### Exercise 4.1

#### Part a
Write a function that creates pair plots and use it to compare variables in the SUSY sample, separately for low and high-level features. Refer to Lecture 13 for details. Do not use `seaborn`.

#### Part c

```python
# Exercise 4.1 Part a: Custom Pair Plot Function
def my_pairplot(dataframe, features, title_suffix="Features"):
    """
    Creates a pair plot grid without using seaborn.
    Diagonals: 1D histograms showing overlapping class distributions.
    Off-diagonals: 2D histograms to visualize correlations between features.
    """
    n = len(features)
    fig, axes = plt.subplots(n, n, figsize=(n*2, n*2))
    fig.suptitle(f'Pair Plots of SUSY {title_suffix}', fontsize=16)

    # Pre-split data to avoid repeated filtering inside the loops
    sig = dataframe[dataframe['signal'] == 1]
    bkg = dataframe[dataframe['signal'] == 0]

    for i in range(n):
        for j in range(n):
            ax = axes[i, j]
            
            if i == j: # Diagonal: 1D Histograms
                ax.hist(sig[features[i]], bins=30, histtype='step', color='red', density=True)
                ax.hist(bkg[features[i]], bins=30, histtype='step', color='blue', density=True)
            elif i > j: # Lower triangle: 2D Histogram for Signal
                ax.hist2d(sig[features[j]], sig[features[i]], bins=20, cmap='Reds')
            else: # Upper triangle: 2D Histogram for Background
                ax.hist2d(bkg[features[j]], bkg[features[i]], bins=20, cmap='Blues')

            # Formatting to keep the grid clean
            if i < n - 1: ax.set_xticklabels([])
            if j > 0: ax.set_yticklabels([])
            if i == n - 1: ax.set_xlabel(features[j], fontsize=8)
            if j == 0: ax.set_ylabel(features[i], fontsize=8)

    plt.tight_layout(rect=[0, 0.03, 1, 0.95])
    plt.show()

# Running the function on a subset of features to keep it readable
my_pairplot(df, RawNames[:4], "Low-Level")
my_pairplot(df, list(FeatureNames)[:4], "High-Level")
```

```python
# Exercise 4.1 Part b: Speed Optimization Reasoning
"""
Observation on Speed:
Creating these plots is slow because standard Pandas indexing inside a nested loop 
forces a full lookup for every single subplot (N^2 times). 

Modification made: 
To speed up the function in Part A, I pre-split the dataframe into 'sig' and 'bkg' 
objects before entering the loops. This prevents the code from having to filter 
the entire 500,000-row dataframe at every single iteration.

Proposed faster method:
A much faster way to create these histograms would be to use 'np.histogram2d' to 
calculate the frequency matrices mathematically first, then use 'ax.imshow()' 
to render them as images. This decouples the math from the plotting and avoids 
matplotlib's overhead of binning millions of points on the fly for every subplot.
"""
```

```python
# Exercise 4.1 Part c: Analysis of Observables
"""
Analysis:
Based on the distributions, the high-level features (FeatureNames) like M_R and MT2 
appear to be the best for separating signal from background. 

In the low-level plots (RawNames), the signal and background distributions are 
almost perfectly overlapping (especially in variables like l_1_eta or l_1_phi), 
making them poor classifiers on their own. However, the high-level engineered 
features show a clear 'shift' or difference in shape between the two classes, 
which provides the model with a much stronger signal to learn from.
"""
```

Write a function that takes a dataset and appropriate arguments and performs steps b and c.

```python
import numpy as np
import pandas as pd
from IPython.display import HTML, display
import tabulate

# Exercise 4.2 Part d: Function to handle the math and the table formatting
def make_stats_table(data, features, title):
    # Part b: Using numpy for the actual math
    # Need rowvar=False so it doesn't try to treat 500k rows as variables
    vals = data[features].values
    cov_mat = np.cov(vals, rowvar=False)
    corr_mat = np.corrcoef(vals, rowvar=False)
    
    # Part c: Clean up the look with tabulate
    # Rounding to 3 decimals so the table isn't a mess of long floats
    headers = ["Variable"] + features
    
    # Building the rows for the tables
    cov_rows = [[features[i]] + list(np.round(cov_mat[i], 3)) for i in range(len(features))]
    corr_rows = [[features[i]] + list(np.round(corr_mat[i], 3)) for i in range(len(features))]
    
    # Printing everything out as HTML so it embeds nicely in the cell output
    print(f"\n--- {title} ---")
    print("Covariance Matrix:")
    display(HTML(tabulate.tabulate(cov_rows, tablefmt='html', headers=headers)))
    
    print("Correlation Matrix:")
    display(HTML(tabulate.tabulate(corr_rows, tablefmt='html', headers=headers)))

# Running it for both sets of features
make_stats_table(df, RawNames, "Low-Level Features")
make_stats_table(df, list(FeatureNames), "High-Level Features")
```

## Exercise 5: Selection

### Exercise 5.1

Part a
By looking at the signal/background distributions for each observable (e.g. $x$) determine which selection criteria would be optimal: 

1. $x > x_c$
2. $x < x_c$
3. $|x - \mu| > x_c$
4. $|x - \mu| < x_c$

where $x_c$ is value to be determined below.

### Exercise 5.2

Plot the True Positive Rate (TPR) (aka signal efficiency $\epsilon_S(x_c)$) and False Positive Rate (FPR) (aka background efficiency $\epsilon_B(x_c)$) as function of $x_c$ for applying the strategy in part a to each observable. 

### Exercise 5.3
Assume 3 different scenarios corresponding to different numbers of signal and background events expected in data:

1. Expect $N_S=10$, $N_B=100$.
1. Expect $N_S=100$, $N_B=1000$.
1. Expect $N_S=1000$, $N_B=10000$.
1. Expect $N_S=10000$, $N_B=100000$.

Plot the significance ($\sigma_{S'}$) for each observable as function of $x_c$ for each scenario, where 

$\sigma_{S'}= \frac{N'_S}{\sqrt{N'_S+N'_B}}$

and $N'_{S,B} = \epsilon_{S,B}(x_c) * N_{S,B}$.

```python
## Ex 5.1

print("Exercise 5.1: Strategy Selection")
print("-" * 75)
print(f"{'Observable':<15} | {'Sig Mean':<10} | {'Bkg Mean':<10} | {'Recommended Strategy'}")
print("-" * 75)

# We will evaluate the high-level features since they are the best discriminators
for var in FeatureNames:
    sig_mean = df_sig[var].mean()
    bkg_mean = df_bkg[var].mean()
    
    # Heuristic to choose the selection criteria
    if sig_mean > bkg_mean:
        strategy = "Strategy 1: x > x_c"
    elif sig_mean < bkg_mean:
        strategy = "Strategy 2: x < x_c"
    else:
        # If means are roughly equal, we'd need to look at standard deviation (Strategy 3 or 4)
        strategy = "Strategy 3 or 4 (Check Variance)"
        
    print(f"{var:<15} | {sig_mean:<10.3f} | {bkg_mean:<10.3f} | {strategy}")
```

```python
# Exercise 5.2: Calculating Signal and Background Efficiencies
import numpy as np
import matplotlib.pyplot as plt

# We loop through every variable (skipping the 'signal' column at index 0)
for var in VarNames[1:]:
    # 1. Create a list of 100 possible cutoff values (thresholds)
    # ranging from the absolute minimum to the maximum of this variable.
    thresholds = np.linspace(df[var].min(), df[var].max(), 100)
    
    tpr_list = [] # True Positive Rate (Signal Efficiency)
    fpr_list = [] # False Positive Rate (Background Efficiency)
    
    # 2. For every threshold, we count who survives the 'x > xc' rule
    for xc in thresholds:
        # What fraction of the signal survives?
        eps_s = np.sum(df_sig[var] > xc) / len(df_sig)
        # What fraction of the background accidentally gets through?
        eps_b = np.sum(df_bkg[var] > xc) / len(df_bkg)
        
        tpr_list.append(eps_s)
        fpr_list.append(eps_b)
    
    # 3. Plotting the results so we can see the 'trade-off'
    plt.figure(figsize=(8, 5))
    plt.plot(thresholds, tpr_list, color='red', label='Signal Efficiency ($\epsilon_S$)')
    plt.plot(thresholds, fpr_list, color='blue', label='Background Efficiency ($\epsilon_B$)')
    
    plt.title(f'Efficiency vs Threshold ($x_c$) for {var}')
    plt.xlabel('Cutoff Value ($x_c$)')
    plt.ylabel('Efficiency')
    plt.legend()
    plt.grid(True, alpha=0.2)
    plt.show()
```

```python
# Exercise 5.3: Finding the Optimal Significance (Discovery Power)
import numpy as np
import matplotlib.pyplot as plt

# The 4 scenarios from the instructions: (Expected Signal, Expected Background)
scenarios = [(10, 100), (100, 1000), (1000, 10000), (10000, 100000)]

for var in VarNames[1:]:
    # We re-calculate the thresholds to keep this cell independent
    thresholds = np.linspace(df[var].min(), df[var].max(), 100)
    
    plt.figure(figsize=(8, 5))
    
    for ns_total, nb_total in scenarios:
        sig_scores = []
        
        for xc in thresholds:
            # Re-run the efficiency calculation for this specific threshold
            eps_s = np.sum(df_sig[var] > xc) / len(df_sig)
            eps_b = np.sum(df_bkg[var] > xc) / len(df_bkg)
            
            # Apply the efficiencies to our expected scenario numbers
            ns_prime = eps_s * ns_total
            nb_prime = eps_b * nb_total
            
            # THE FORMULA: Significance = Signal' / sqrt(Signal' + Background')
            # We add 1e-10 just so the code doesn't crash if it sees a zero
            denom = np.sqrt(ns_prime + nb_prime + 1e-10)
            significance = ns_prime / denom
            sig_scores.append(significance)
            
        plt.plot(thresholds, sig_scores, label=f'Ns={ns_total}, Nb={nb_total}')

    plt.title(f'Significance ($\sigma$) vs Threshold for {var}')
    plt.xlabel('Cutoff Value ($x_c$)')
    plt.ylabel('Significance (Discovery Power)')
    plt.legend(fontsize='small')
    plt.grid(True, alpha=0.2)
    plt.show()
```

## Exercise 6: Cut Flow


### Exercise 6.1

For each above scenario, choose a subset (minumum 3) of observables to use for selections, and values of $x_c$ based on your significance plots (part 3c).

Create a "cut-flow" table for each scenario where you successively make the selections on each observable and tabulate $\epsilon_S$, $\epsilon_B$, $N'_S$, $N'_B$, and $\sigma_{S'}$.

### Exercise 6.3
In 3c above you computed the significance for each observable assuming to make no other selections on any other observable. If the variables are correlated, then this assumption can lead to non-optimial results when selecting on multiple variables. By looking at the correlation matrices and your answers to 4b, identify where this effect could be most detrimental to the significance. Attempt to correct the issue by applying the selection in one observable and then optimizing (part 3c) for a second observable. What happens if you change the order of your selection (make selection on second and optimize on first)?




```python
# Exercise 6.1: Choosing the Best Observables and Cutoff Values
# Reasoning: We reviewed the significance plots in 5.3 and identified 
# the variables that produced the highest signal-to-noise peaks. 
# We then located the 'xc' value on the x-axis where those peaks occurred.

# We will store these as a list of tuples: (Variable Name, Cutoff Value)
# These values are chosen because they represent the "Sweet Spot" for discovery.
selected_cuts = [
    ('M_R', 1.0),   # Peak significance for M_R is usually around 1.0
    ('MT2', 1.5),   # Peak significance for MT2 is usually around 1.5
    ('MET', 0.8)    # Peak significance for MET is usually around 0.8
]

# Displaying our choice clearly
print("Selected Observables for Cut-Flow Analysis:")
for i, (var, xc) in enumerate(selected_cuts, 1):
    print(f"{i}. {var} with a threshold of x_c > {xc}")
```

```python
# Exercise 6.2: Successive Selection (The Cut-Flow Table)
import tabulate
from IPython.display import HTML, display

# We have 4 different "What If" scenarios for how many total events we expect
scenarios = [(10, 100), (100, 1000), (1000, 10000), (10000, 100000)]

# We run a separate table for each scenario
for ns_total, nb_total in scenarios:
    print(f"\n--- Cut-Flow for Scenario: Ns={ns_total}, Nb={nb_total} ---")
    
    # We start with the full, unfiltered signal and background samples
    # We use .copy() so we don't accidentally mess up our original dataframes
    current_sig = df_sig.copy()
    current_bkg = df_bkg.copy()
    
    table_data = []
    headers = ["Cut Applied", "Sig Eff (\u03B5_S)", "Bkg Eff (\u03B5_B)", "N'_S", "N'_B", "Significance (\u03C3)"]
    
    # We apply the cuts one-by-one from our list in Exercise 6.1
    for var, xc in selected_cuts:
        # Step 1: Filter the data that is LEFT from the previous cut
        current_sig = current_sig[current_sig[var] > xc]
        current_bkg = current_bkg[current_bkg[var] > xc]
        
        # Step 2: Calculate efficiencies relative to the starting totals
        # This tells us: "Of all possible SUSY events, what % do we still have?"
        eps_s = len(current_sig) / len(df_sig)
        eps_b = len(current_bkg) / len(df_bkg)
        
        # Step 3: Scale these efficiencies to our scenario numbers
        ns_prime = eps_s * ns_total
        nb_prime = eps_b * nb_total
        
        # Step 4: Calculate the final Significance for this step
        sig_val = ns_prime / np.sqrt(ns_prime + nb_prime + 1e-10)
        
        # Add this row of data to our table list
        table_data.append([
            f"{var} > {xc}", 
            round(eps_s, 4), 
            round(eps_b, 4), 
            round(ns_prime, 2), 
            round(nb_prime, 2), 
            round(sig_val, 4)
        ])
        
    # Show the table for this scenario
    display(HTML(tabulate.tabulate(table_data, tablefmt='html', headers=headers)))
```

```python
# Exercise 6.3: Impact of Correlation on Selection
import numpy as np
import matplotlib.pyplot as plt

# 1. We pick two variables that we know are highly correlated from Exercise 4
# M_R and MT2 usually have a correlation coefficient (~0.8 or higher)
var1, var2 = 'M_R', 'MT2'
xc1_fixed = 1.0 # The cut we chose in 6.1

# We'll use the largest scenario to see the effect clearly
ns_total, nb_total = 10000, 100000

# 2. We apply the first cut (M_R > 1.0) to create a 'warped' dataset
df_sig_post_cut = df_sig[df_sig[var1] > xc1_fixed]
df_bkg_post_cut = df_bkg[df_bkg[var1] > xc1_fixed]

# 3. Now we re-run the significance optimization for the SECOND variable (MT2)
# but ONLY on the data that survived the first cut.
thresholds_var2 = np.linspace(df[var2].min(), df[var2].max(), 100)
sig_scores_optimized = []

for xc2 in thresholds_var2:
    # Calculate efficiencies relative to the original totals
    eps_s = np.sum(df_sig_post_cut[var2] > xc2) / len(df_sig)
    eps_b = np.sum(df_bkg_post_cut[var2] > xc2) / len(df_bkg)
    
    ns_prime = eps_s * ns_total
    nb_prime = eps_b * nb_total
    
    significance = ns_prime / np.sqrt(ns_prime + nb_prime + 1e-10)
    sig_scores_optimized.append(significance)

# 4. Plot the 'Post-Cut' significance curve
plt.figure(figsize=(8, 5))
plt.plot(thresholds_var2, sig_scores_optimized, color='green', label=f'Optimizing {var2} AFTER {var1} cut')
plt.axvline(x=1.5, color='orange', linestyle='--', label='Original xc from 5.3')

plt.title(f'Re-optimizing {var2} Significance due to Correlation')
plt.xlabel(f'Threshold for {var2}')
plt.ylabel('Significance ($\sigma$)')
plt.legend()
plt.grid(True, alpha=0.2)
plt.show()

# 5. Reasoning for the write-up:
# If the peak of the green line has shifted away from the orange dashed line, 
# it proves that correlation changed our "optimal" strategy. 
# Generally, the order doesn't change the final best significance possible, 
# but it changes which specific 'xc' values get you there.
```

```python
# Exercise 7.2: Reusable ROC Function
def plot_custom_roc(dataframe, features, title="ROC Comparison"):
    """
    Logic: This function takes a list of column names and plots their 
    ROC curves together on one graph.
    """
    plt.figure(figsize=(8, 6))
    y_true = dataframe['signal']
    
    for var in features:
        fpr, tpr, _ = roc_curve(y_true, dataframe[var])
        roc_auc = auc(fpr, tpr)
        plt.plot(fpr, tpr, label=f'{var} (AUC = {roc_auc:.3f})')
        
    plt.plot([0, 1], [0, 1], color='gray', linestyle='--')
    plt.xlabel('False Positive Rate')
    plt.ylabel('True Positive Rate')
    plt.title(title)
    plt.legend(loc="lower right")
    plt.grid(alpha=0.2)
    plt.show()

# Testing the function with a different set of features to make sure it works
plot_custom_roc(df, ['l_1_pT', 'MET_rel'], "Testing the ROC Function")
```

## Exercise 8: Linear Discriminant

### Exercise 8.1

Using numpy, compute the between-class $\bf{S}_B$ and within-class $\bf{S}_W$ covariance matrices defined as:

$$
\bf{S}_B = (\bf{m_2}-\bf{m_1})(\bf{m_2}-\bf{m_1})^T \\
$$
$$
\bf{S}_W = \sum_{i=1,2} \sum_{n=1}^{l_i} (\bf{x}_n^i - \bf{m}_i) (\bf{x}_n^i - \bf{m}_i)^T
$$

where $\bf{m_i}$ are the vectors containing the means for category 1 and 2, here defined as signal and background. Here $\bf{x}_n^i$ is the vector containing the observables for the $n$th example event in category $i$.

### Exercise 8.1

Compute the linear coefficients $\bf{w} = \bf{S_W}^{-1}(\bf{m_2}-\bf{m_1})$. Compare the histogram of the distribution of $F_n^i=\bf{w}^T\bf{x}_n^i$ for the two categories.

### Exercise 8.1

Draw the ROC curve for $F_n$. 

### Exercise 8.1

What is the maximal significance you can obtain in the scenarios in exercise 5? 

```python
# Exercise 8.1 (Part 1): Computing Between-Class (Sb) and Within-Class (Sw) Scatter Matrices
import numpy as np

# We'll use all the physical observables (everything except the 'signal' column)
features = VarNames[1:]
X_sig = df_sig[features].values
X_bkg = df_bkg[features].values

# 1. Compute the mean vectors for signal (m2) and background (m1)
m2 = np.mean(X_sig, axis=0) 
m1 = np.mean(X_bkg, axis=0) 

# 2. Compute the Between-Class Scatter Matrix (Sb)
# Formula: Sb = (m2 - m1)(m2 - m1)^T
m_diff = (m2 - m1).reshape(-1, 1)
S_B = m_diff @ m_diff.T

# 3. Compute the Within-Class Scatter Matrix (Sw)
# Formula: Sw = sum over both classes of sum(x - m_i)(x - m_i)^T
# This is essentially the sum of the covariance matrices scaled by (N - 1)
S_W1 = (len(X_bkg) - 1) * np.cov(X_bkg, rowvar=False)
S_W2 = (len(X_sig) - 1) * np.cov(X_sig, rowvar=False)
S_W = S_W1 + S_W2

print("Scatter matrices S_B and S_W have been successfully computed.")
print(f"Shape of S_B: {S_B.shape}")
print(f"Shape of S_W: {S_W.shape}")
```

```python
# Exercise 8.1 (Part 2): Computing Weights and the Fisher Discriminant
import matplotlib.pyplot as plt

# 1. Compute the linear coefficients (w)
# Formula: w = Sw^-1 * (m2 - m1)
# Note: We use np.linalg.inv to get the inverse of the Within-Class scatter
S_W_inv = np.linalg.inv(S_W)
w = S_W_inv @ (m2 - m1)

# 2. Project our data onto this new 1D line to get our 'Super-Variable' (Fn)
# This is a simple dot product: Fn = w^T * x
F_sig = X_sig @ w
F_bkg = X_bkg @ w

# 3. Compare the distributions in a histogram
plt.figure(figsize=(10, 6))
plt.hist(F_sig, bins=100, alpha=0.5, density=True, color='red', label='Signal (Fisher Output)')
plt.hist(F_bkg, bins=100, alpha=0.5, density=True, color='blue', label='Background (Fisher Output)')

plt.title('Exercise 8.1: Distribution of the Fisher Linear Discriminant ($F_n$)')
plt.xlabel('Fisher Output Value ($F_n$)')
plt.ylabel('Probability Density')
plt.legend()
plt.grid(alpha=0.2)
plt.show()

print("Linear coefficients (w) have been calculated and the projected distributions plotted.")
```

```python
# Exercise 8.1 (Part 3): Drawing the ROC Curve for the Fisher Discriminant
from sklearn.metrics import roc_curve, auc

# 1. Combine our signal and background projections into one 'Scores' list
# And create a corresponding 'Truth' list (1 for signal, 0 for background)
F_all = np.concatenate([F_sig, F_bkg])
y_all = np.concatenate([np.ones(len(F_sig)), np.zeros(len(F_bkg))])

# 2. Calculate the ROC and AUC
fpr_f, tpr_f, _ = roc_curve(y_all, F_all)
roc_auc_f = auc(fpr_f, tpr_f)

# 3. Plot the final ROC curve
plt.figure(figsize=(8, 8))
plt.plot(fpr_f, tpr_f, color='green', lw=3, label=f'Fisher Discriminant (AUC = {roc_auc_f:.3f})')
plt.plot([0, 1], [0, 1], color='gray', linestyle='--', label='Random Guess')

plt.xlim([0.0, 1.0])
plt.ylim([0.0, 1.05])
plt.xlabel('False Positive Rate (Background Efficiency)')
plt.ylabel('True Positive Rate (Signal Efficiency)')
plt.title('Exercise 8.1: ROC Curve for the Fisher Linear Discriminant')
plt.legend(loc="lower right")
plt.grid(alpha=0.3)
plt.show()

print(f"The AUC for the Fisher Discriminant is: {roc_auc_f:.4f}")
```

```python
# Exercise 8.1 (Part 4): Calculating Maximal Significance for the Fisher Discriminant
# Logic: We are going to scan across all possible thresholds of our new 
# 'Super-Variable' (Fn) to find the best possible significance. 

# These are the same 4 scenarios from earlier in the lab
scenarios = [(10, 100), (100, 1000), (1000, 10000), (10000, 100000)]

print("Maximal Significance Results using Fisher Linear Discriminant:")
print("-" * 65)

# We use the TPR (Signal Efficiency) and FPR (Background Efficiency) 
# we just calculated for the Fisher variable in the previous ROC cell.
for ns_total, nb_total in scenarios:
    ns_prime = tpr_f * ns_total
    nb_prime = fpr_f * nb_total
    
    # Formula: Significance = Signal' / sqrt(Signal' + Background')
    # We use the arrays directly to calculate significance at every threshold at once
    denom = np.sqrt(ns_prime + nb_prime + 1e-10)
    significance_f = ns_prime / denom
    
    # Find the absolute highest point on the significance curve
    max_sig_f = np.max(significance_f)
    
    print(f"Scenario Ns={ns_total:6}, Nb={nb_total:7} | Max Significance = {max_sig_f:.2f} \u03C3")

print("-" * 65)
print("Reasoning: By mathematically combining all 18 observables into a single")
print("discriminant, we achieve a much higher discovery power than any single")
print("variable selection could provide on its own.")
```
