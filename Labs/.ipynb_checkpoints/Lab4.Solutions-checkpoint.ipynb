1. Write a "counter" class that can be incremented up to a specified maximum value, will print an error if an attempt is made to increment beyond that value, and allows reseting the counter.

```python
class counter:
    def __init__(self,max_counts=10):
        self.__n_counts = 0
        self.__max_counts=max_counts

    def n_counts(self):
        return self.__n_counts

    def add_counts(self,n=1):
        if isinstance(n,int):
            self.__n_counts+=n
            if self.__n_counts>self.__max_counts:
                self.__n_counts=self.__max_counts
                print ("Error. Reached max value.")
        else:
            print ("Number of passengers must be an integer.")

    def reset(self):
        self.__n_counts = 0
```

```python
count_test = counter(5)

count_test.add_counts(2)
print (count_test.n_counts())
```

```python
count_test = counter(5)

count_test.add_counts(6)
print (count_test.n_counts())
```

```python
count_test = counter(5)

count_test.reset()
print(count_test.n_counts())
```

2. Copy and paste your solution to question 1 and modify it so that all the data held by the counter is private. Implement functions to check the value of the counter, check the maximum value, and check if the counter is at the maximum.

```python
class counter:
    def __init__(self,max_counts=10):
        self.__n_counts = 0
        self.__max_counts=max_counts

    def n_counts(self):
        return self.__n_counts
    def max_counts(self):
        return self.__max_counts

    def add_counts(self,n=1):
        if isinstance(n,int):
            self.__n_counts+=n
            if self.__n_counts>self.__max_counts:
                self.__n_counts=self.__max_counts
                print ("Error. Reached max value.")
        else:
            print ("Number of passengers must be an integer.")
            
    def is_at_max(self):
        return self.__n_counts == self.__max_counts

    def reset(self):
        self.__n_counts = 0
```

```python
count_test = counter(3)

print(count_test.n_counts())
print(count_test.max_counts())
print(count_test.is_at_max())
```

```python
count_test = counter(3)
count_test.add_counts(1)

print(count_test.n_counts())
print(count_test.max_counts())
print(count_test.is_at_max())
```

```python
count_test = counter(3)
count_test.add_counts(7)

print(count_test.n_counts())
print(count_test.max_counts())
print(count_test.is_at_max())
```

3. Implement a class to represent a rectangle, holding the length, width,and coordinates of a corner of the object. Implement functions that compute the area and perimeter of the rectangle. Make all data members private and privide accessors to retrieve values of data members.

```python
class rectangle:
    def __init__(self, width, length, x, y):
        self.__width = width
        self.__length = length
        self.__x = x
        self.__y = y

    def n_length(self):
        return self.__length
        
    def n_width(self):
        return self.__width
        
    def x_coord(self):
        return self.__x
        
    def y_coord(self):
        return self.__y

    def find_coords(self):
        return (self.__x,self.__y)
        
    def area(self):
        return self.__length * self.__width
        
    def perimeter(self):
        return ((2 * self.__length) + (2 * self.__width))
```

```python
rectangle_test = rectangle(3,4,2,2)


print(rectangle_test.find_coords())
print(rectangle_test.area())
print(rectangle_test.perimeter())
```

4. Implement a class to represent a circle, holding the radius x and y coordinates of center of the object. Implement functions that compute the area and perimeter of the circle. Make all data members private and privide accessors to retrieve values of data members.

```python
class circle:
    def __init__(self, radius, x, y):
        self.__radius = radius
        self.__x = x
        self.__y = y

    def radius(self):
        return self.__radius

    def x_coord(self):
        return self.__x
        
    def y_coord(self):
        return self.__y
        
    def find_coords(self):
        return (self.__x,self.__y)
        
    def area(self):
        return (self.__radius **2) * math.pi

    def perimeter(self):
        return 2 * math.pi * self.__radius
```

```python
circle_test = circle(3,2,2)

print(circle_test.find_coords())
print(circle_test.area())
print(circle_test.perimeter())
```

5. Implement a common base class for the classes implemented in 3 and 4 above which implements all common methods as not implemented functions (virtual). Re-implement your regtangle and circule classes to inherit from the base class and overload the functions accordingly.

```python
class common_base:
    def __init__(self, x, y):
        self.__x = x
        self.__y = y

    def get_x(self):
        return self.__x

    def get_y(self):
        return self.__y

    def find_coords(self):
        raise NotImplementedError

class rectangle(common_base):
    def __init__(self, x, y, width, length):
        common_base.__init__(self, x, y)
        self.__width = width
        self.__length = length

    def area_rectangle(self):
        return self.__length * self.__width

    def perimeter_rectangle(self):
        return ((2 * self.__length) + (2 * self.__width))

    def find_coords(self):
        print(self.get_x(), self.get_y())

class circle(common_base):
    def __init__(self, x, y, radius):
        common_base.__init__(self, x, y)
        self.__radius = radius

    def area_circle(self):
        return (self.__radius **2) * math.pi

    def perimeter_circle(self):
        return 2 * math.pi * self.__radius

    def find_coords(self):
        print(self.get_x(), self.get_y())
```

```python
a_rectangle=rectangle(2,3,4,5)
a_circle=circle(6,7,8)

a_rectangle.find_coords()
a_circle.find_coords()

print(a_rectangle.area_rectangle())
print(a_rectangle.perimeter_rectangle())

print(a_circle.area_circle())
print(a_circle.perimeter_circle())
```

6. Implement a triangle class analogous to the rectangle and circle in question 5.

```python
class triangle_coordinates:
    def __init__(self, x1, y1, x2, y2, x3, y3):
        self.__x1 = x1
        self.__y1 = y1
        self.__x2 = x2
        self.__y2 = y2
        self.__x3 = x3
        self.__y3 = y3

    def get_x1(self):
        return self.__x1

    def get_y1(self):
        return self.__y1

    def get_x2(self):
        return self.__x2

    def get_y2(self):
        return self.__y2

    def get_x3(self):
        return self.__x3

    def get_y3(self):
        return self.__y3

    def show_coords(self):
        raise NotImplementedError

class triangle(triangle_coordinates):
    def __init__(self, x1, y1, x2, y2, x3, y3):
        triangle_coordinates.__init__(self, x1,y1,x2,y2,x3,y3)

    def side_1(self):
        return math.sqrt(((self.get_x2() - self.get_x1()) ** 2) + ((self.get_y2() - self.get_y1()) ** 2))
        
    def side_2(self):
        return math.sqrt(((self.get_x3() - self.get_x2()) ** 2) + ((self.get_y3() - self.get_y2()) ** 2))
    
    def side_3(self):
        return math.sqrt(((self.get_x1() - self.get_x3()) ** 2) + ((self.get_y1() - self.get_y3()) ** 2))

    def mid_point_side_3_x(self):
        return ((self.get_x1() + self.get_x3())* 0.5)

    def mid_point_side_3_y(self):
        return ((self.get_y1() + self.get_y3())* 0.5)

    def height_triangle(self):
        return math.sqrt(((self.mid_point_side_3_x() - self.get_x1())**2 + ((self.mid_point_side_3_y() - self.get_y1()) **2)))

    def area_triangle(self):
        return (0.5 * self.side_1() * self.height_triangle())

    def perimeter_triangle(self):
        return self.side_1() + self.side_2() + self.side_3()

    def show_coordinates(self):
        print((self.get_x1(), self.get_y1()),(self.get_x2(), self.get_y2()),(self.get_x3(), self.get_y3()))
```

```python
test_triangle = triangle(0,0,1,1,-1,-1)

print(test_triangle.side_1())
print(test_triangle.side_2())
print(test_triangle.side_3())

print(test_triangle.height_triangle())
print(test_triangle.area_triangle())
print(test_triangle.perimeter_triangle())

print(test_triangle.show_coordinates())
```
