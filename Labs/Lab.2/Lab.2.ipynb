*Exercise 4:* Write a function that takes an `n×n` matrix representing a tic‑tac‑toe game and returns one of the following values:

- `-1` if the game is **incomplete** (still empty spaces and no winner)
- `0` if the game is a **draw**
- `1` if **player 1** (`"X"`) has won
- `2` if **player 2** (`"O"`) has won

Here are some example inputs you can use to test your code:


````python
# Write your solution 
def check_play(board):
    r = len(board)
    c = len(board[0])

    for i in range(r):
            if all(board[i][j] == 1 for j in range(c)):
                return 1
            elif all(board[i][j] == 2 for j in range(c)):
                return 2
                
    for j in range(c):
         if all(board[i][j] == 1 for i in range(r)):
                return 1
         elif all(board[i][j] == 2 for i in range(r)):
                return 2
             
        
    if all(board[i][i] == 1 for i in range(r)):
        return 1
    elif all(board[i][i] == 2 for i in range(r)):
        return 2

    if all(board[i][r-1-i] == 1 for i in range(r)):
        return 1
    elif all(board[i][r-1-i] == 2 for i in range(r)):
        return 2


    for i in range(r):
        if any(board[i][j] == 0 for j in range(c)):
            return -1
            
    return 0
````

````python
# Write your solution here
def player_move(board, player, x, y):
    col = ord(x.upper()) - ord('A')
    row = int(y) - 1

    if board[row][col] != 0:
        return "Invalid move!"

    board[row][col] = player
    return True
````

*Exercise 11:* Develop a version of the game where one player is the computer.

Note: you do **not** need an extensive search for the best move. For example, you can have the computer:
- block obvious losses
- otherwise try to create a winning row/column/diagonal


````python
# Write your solution here
import random

def computer_move(board, computer, human):
    r = len(board)
    c = len(board[0])

    def try_move(player):
        for i in range(r):
            for j in range(c):
                if board[i][j] == 0:
                    board[i][j] = player
                    if check_play(board) == player:
                        return (i, j)
                    board[i][j] = 0
        return 

    move = try_move(computer)
    if move:
        board[move[0]][move[1]] = computer
        return
    move = try_move(human)
    if move:
        board[move[0]][move[1]] = computer
        return
    empty = [(i, j) for i in range(r) for j in range(c) if board[i][j] == 0]
    if empty:
        i, j = random.choice(empty)
        board[i][j] = computer
````

````python
# Test your solution here
x_size = int(input("Enter the number of rows on board: "))
y_size = int(input("Enter the number of columns on board: "))

b = make_board(x_size, y_size)
draw_board(b)

mode = int(input("Press 1 for 2-player game, 2 to play vs computer: "))
player = 1
computer = 2 if mode == 2 else None
moves = -1

while moves == -1:

    if mode == 2 and player == computer:
        print("Computer is making a move...")
        computer_move(b, computer, 1)
    else:
        print("Enter coordinates to make the move")
        x1 = input("Enter the column (A, B, C...): ")
        x2 = input("Enter the row (1, 2, 3...): ")

        result = player_move(b, player, x1, x2)
        if result != True:
            print(result)
            continue

    draw_board(b)
    moves = check_play(b)

    if moves == -1:
        player = 2 if player == 1 else 1

if moves == 1:
    print("X won.")
elif moves == 2:
    print("O won.")
else:
    print("Its a draw.")
````

*Exercise 12:* Develop a version of the game where one player is the computer. This time, write a computer player using exhaustive search with a max depth parameter, similar to lecture.

````python
# Write your solution here
import random

def minimax(board, depth, max_depth, player, smart, dumb):
    result = check_play(board)

    if result == smart:
        return 100 - depth
    if result == dumb:
        return depth - 100
    if result == 0:
        return 0

    if depth == max_depth:
        return 0

    r = len(board)
    c = len(board[0])

    if player == smart:
        best = -float("inf")

        for i in range(r):
            for j in range(c):
                if board[i][j] == 0:
                    board[i][j] = smart
                    score = minimax(board, depth + 1, max_depth,
                                    dumb, smart, dumb)
                    board[i][j] = 0
                    best = max(best, score)

        return best

    else:
        best = float("inf")

        for i in range(r):
            for j in range(c):
                if board[i][j] == 0:
                    board[i][j] = dumb
                    score = minimax(board, depth + 1, max_depth,
                                    smart, smart, dumb)
                    board[i][j] = 0
                    best = min(best, score)

        return best


def computer_move_exhaustive(board, player, opponent, depth):
    r = len(board)
    c = len(board[0])

    best_score = -float("inf")
    best_move = None

    for i in range(r):
        for j in range(c):
            if board[i][j] == 0:
                board[i][j] = player
                score = minimax(board, 0, depth,
                                opponent, player, opponent)
                board[i][j] = 0

                if score > best_score:
                    best_score = score
                    best_move = (i, j)

    if best_move:
        board[best_move[0]][best_move[1]] = player
````

*Exercise 13:* Make the 2 computer players play each-other for 10 games on a 3x3, then 4x4, then 5x5 grid. Set the max depth so that the games only take seconds. Measure the "smarter" player's win rate for each grid.

````python
# Write your solution 
def play_ai_vs_ai(size, depth_smart, depth_dumb, games=10):
    smart_wins = 0
    draw= 0
    for g in range(games):
        board = make_board(size, size)

        smart = 1 if g % 2 == 0 else 2
        dumb = 2 if smart == 1 else 1

        player = 1
        result = -1

        while result == -1:
            if player == smart:
                computer_move_exhaustive(board, smart, dumb, depth_smart)
            else:
                computer_move_exhaustive(board, dumb, smart, depth_dumb)

            result = check_play(board)
            player = 2 if player == 1 else 1

        if result == smart:
            smart_wins += 1
        elif result == 0:
            draw += 1

    win_rate = smart_wins / games * 100
    print(f"{size}x{size} grid → Smart AI win rate: {win_rate:.1f}% with {draw} draws.")
````
